```python
import pandas as pd
df = pd.read_csv("data.csv")
print(df.groupby("Type1")["Weight"].mean())
```

```python
avg = df.groupby("Type1")["Weight"].mean().sort_values(ascending=False)
avg.plot(kind = "bar", title = "Average weight by Pokemon type")
```

**WHICH POKEMON TYPE IS THE HEAVIEST ON AVERAGE?**

I used a Pokemon dataset of 150 Pokemon and grouped them by their main type (Type1).

**FINDINGS:**
- Rock is the heaviest type on average (about 87.6), followed by Dragon (76.6) and Psychic (58.4).
- Ghost is the lightest (about 13.6).
- The heavy types are mostly tough, large Pokemon, while light types like Ghost and Bug are small.

**CONCLUSION:**

Rock type Pokemon are the heaviest on average, and Ghost type are the lightest.
